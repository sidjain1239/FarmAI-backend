```python
 # Import libraries
import pandas as pd
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
#  Load the dataset
data = pd.read_csv('fertilizer_recommendation_dataset.csv')

# Clean the column: remove 'c' and make lowercase
data['Soil'] = data['Soil'].str.replace('Soil', '', case=False).str.lower().str.strip()
print(data)
```

```python
# Separate features and target
X = data.drop(['Fertilizer', 'Remark'], axis=1)
y = data['Fertilizer']

#  Label encode all categorical features
categorical_cols = ['Soil', 'Crop']  # only these two
encoders = {}

for col in categorical_cols:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col])
    encoders[col] = le  

#  Label encode target Fertilizer
fertilizer_encoder = LabelEncoder()
y_encoded = fertilizer_encoder.fit_transform(y)
```

```python
# Split the data
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

# Train the model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)
```

```python
#  Save model and encoders
joblib.dump(model, 'fertilizer_model_label_encoded.pkl')
joblib.dump(encoders, 'fertilizer_feature_encoders.pkl')
joblib.dump(fertilizer_encoder, 'fertilizer_target_encoder.pkl')
joblib.dump(X.columns.tolist(), 'fertilizer_feature_columns.pkl')

# Evaluate model
y_pred = model.predict(X_test)
acc = accuracy_score(y_test, y_pred)
print(f"✅ Model Accuracy: {acc * 100:.2f}%")

#  Prepare Fertilizer-Remark Mapping
fertilizer_remark_mapping = dict(zip(data['Fertilizer'], data['Remark']))
joblib.dump(fertilizer_remark_mapping, 'fertilizer_remark_mapping.pkl')
```
